## 1. Create Dataset

```python
import pandas as pd

data = {
    "label": ["Spam", "Ham", "Spam", "Ham", "Spam", "Ham", "Spam", "Ham", "Spam", "Ham"],
    "message": [
        "Congratulations! You have won a free prize.",
        "Your assignment deadline is Friday. Please submit it on time.",
        "Get a free mobile phone now.",
        "Your class will start tomorrow at 9 AM.",
        "Claim your free gift card before the offer expires.",
        "Please find the lecture notes attached.",
        "You have won a cash lottery.",
        "Thank you for attending the meeting.",
        "Congratulations! Claim your free reward now.",
        "Please share your feedback."
    ]
}

df = pd.DataFrame(data)
display(df)
```

## 2. EDA (Explore the Dataset)

```python
display(df.head())   # First 5 rows
print("Shape:", df.shape)   # Rows and columns
df.info()   # Column information
```

## 3. Text Preprocessing

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

# Make text lowercase
df['lower_text'] = df['message'].str.lower()

# Split each message into words (tokens)
df['tokens'] = df['lower_text'].apply(word_tokenize)

# Remove stop words and punctuation
stop_words = set(stopwords.words('english'))
df['clean_tokens'] = df['tokens'].apply(
    lambda words: [word for word in words if word.isalpha() and word not in stop_words]
)

display(df[['message', 'lower_text', 'tokens', 'clean_tokens']])
```

## 4. Stemming and Lemmatization

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Stemming: shorten words to their stem
df['stemmed'] = df['clean_tokens'].apply(
    lambda words: [stemmer.stem(word) for word in words]
)

# Lemmatization: convert words to dictionary form
df['lemmatized'] = df['clean_tokens'].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)

display(df[['message', 'stemmed', 'lemmatized']])
```

## 5. Bag of Words (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

# Join words back into a sentence
df['processed_text'] = df['lemmatized'].apply(lambda words: ' '.join(words))

# Count how many times each word appears
vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(df['processed_text'])

print("Words in vocabulary:")
print(vectorizer.get_feature_names_out())

print("BoW matrix:")
display(pd.DataFrame(bow.toarray(), columns=vectorizer.get_feature_names_out()))
```

## 6. Vector Embeddings

This model converts each email message into a numerical vector.
