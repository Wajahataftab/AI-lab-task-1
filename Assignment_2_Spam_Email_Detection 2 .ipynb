```python
# Text Preprocessing: Lowercasing, Tokenization, Stop Word Removal

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

# 1. Lowercasing
df["lower_text"] = df["message"].str.lower()

# 2. Tokenization
df["tokens"] = df["lower_text"].apply(word_tokenize)

# 3. Stop Word Removal
stop_words = set(stopwords.words("english"))

df["tokens_no_stopwords"] = df["tokens"].apply(
    lambda tokens: [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]
)

# Display the results
display(
    df[["message", "lower_text", "tokens", "tokens_no_stopwords"]]
)
```

```python
# Step 4: Stemming and Step 5: Lemmatization

from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Apply stemming
df["stemmed_words"] = df["tokens_no_stopwords"].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)

# Apply lemmatization
df["lemmatized_words"] = df["tokens_no_stopwords"].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)

# Display results
display(
    df[[
        "message",
        "tokens_no_stopwords",
        "stemmed_words",
        "lemmatized_words"
    ]]
)
```

```python
# Bag of Words (BoW) Feature Extraction

from sklearn.feature_extraction.text import CountVectorizer

# Convert cleaned tokens back into sentences
df["processed_text"] = df["lemmatized_words"].apply(
    lambda words: " ".join(words)
)

# Create Bag of Words vectors
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(df["processed_text"])

# Display vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Display numerical matrix
bow_df = pd.DataFrame(
    X_bow.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\nBag of Words Matrix:")
display(bow_df)
```

```python
# Step 6: Generate Semantic Vector Embeddings

from sentence_transformers import SentenceTransformer
import pandas as pd

# Load a pretrained embedding model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# Use the original email messages
messages = df["message"].tolist()

# Generate one embedding vector for each email
embeddings = model.encode(messages)

# Display the embedding matrix dimensions
print("Number of email messages:", len(messages))
print("Embedding matrix shape:", embeddings.shape)

# Display the first email and its embedding
print("\nFirst email:")
print(messages[0])

print("\nFirst 10 values of its embedding:")
print(embeddings[0][:10])

# Display a preview of the embedding matrix
embedding_df = pd.DataFrame(
    embeddings,
    index=[f"Email {i+1}" for i in range(len(messages))]
)

print("\nEmbedding matrix preview:")
display(embedding_df.iloc[:, :10])
```

```python
# Save all embedding values in a CSV file
embedding_results = pd.DataFrame(embeddings)
embedding_results.insert(0, "label", df["label"])
embedding_results.insert(1, "message", df["message"])

embedding_results.to_csv("email_embeddings.csv", index=False)

print("Embedding results saved successfully.")
display(embedding_results.head())
```
