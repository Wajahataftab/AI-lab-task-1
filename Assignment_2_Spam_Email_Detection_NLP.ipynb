```python
import pandas as pd
emails = [
    "Congratulations! You have won a free prize. Click here to claim your reward.",
    "Get a free mobile phone now. Claim your free gift card before the offer expires.",
    "You have won a cash lottery. Click now to receive your money.",
    "Limited offer! Claim your free reward today. You are a lucky winner.",
    "Congratulations, you have been selected to receive a free gift.",
    "Your assignment deadline is Friday. Please submit it on time.",
    "Your class will start tomorrow at 9 AM.",
    "Please find the lecture notes attached.",
    "Thank you for attending the meeting. Please share your feedback.",
    "The project meeting has been moved to Monday. Please review the notes."
]
labels = ["Spam", "Spam", "Spam", "Spam", "Spam",
          "Ham", "Ham", "Ham", "Ham", "Ham"]
df= pd.DataFrame({"email": emails, "label": labels})
display(df)
```

```python

display(df.head())
print("Shape:", df.shape)
df.info()
```

```python

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
df["lowercase"] = df["email"].str.lower()
df["clean_text"] = df["lowercase"].apply(lambda x: re.sub(r"[^a-z\s]", " ", x))
# Tokenization: split each email into words
df["tokens"] = df["clean_text"].apply(lambda x: x.split())
df["without_stopwords"] = df["tokens"].apply(
    lambda words: [word for word in words if word not in stop_words]
)
df["stemmed"] = df["without_stopwords"].apply(
    lambda words: [stemmer.stem(word) for word in words]
)
df["lemmatized"] = df["without_stopwords"].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)
display(df[["email", "lowercase", "tokens", "without_stopwords", "stemmed", "lemmatized"]])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
df["processed_text"] = df["lemmatized"].apply(lambda words: " ".join(words))
bow = CountVectorizer()
bow_matrix = bow.fit_transform(df["processed_text"])
bow_result = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow.get_feature_names_out()
)
print("Bag of Words result:")
display(bow_result)
print("BoW shape:", bow_matrix.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df["processed_text"])
tfidf_result = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)
print("TF-IDF vector representations:")
display(tfidf_result.round(3))
print("TF-IDF shape:", tfidf_matrix.shape)
```

```python
print("Email labels:")
display(df[["email", "label"]])
print("Number of Spam and Ham emails:")
display(df["label"].value_counts().to_frame("Count"))
print("Preprocessing results:")
display(df[["email", "without_stopwords", "stemmed", "lemmatized"]])
print("BoW matrix shape:", bow_matrix.shape)
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```
